##Anomalies/Outliers Time-Series
Do ponto de vista tradicional, um outlier (ou anomalia) é: **“uma observação que se desvia tanto das demais observações a ponto de levantar suspeitas de que foi gerada por um mecanismo diferente.”** Portanto, podemos considerar outliers como observações que não seguem o comportamento esperado.

Um **outlier pontual (point outlier)** é um dado que apresenta um comportamento incomum em um instante específico do tempo quando comparado aos demais valores da série temporal (**outlier global**) ou aos seus pontos vizinhos (**outlier local**). Os outliers pontuais podem ser **univariados** ou **multivariados**, dependendo de afetarem uma única variável dependente do tempo ou múltiplas variáveis dependentes do tempo, respectivamente.

**Outlier de subsequência (Subsequence outlier)** refere-se a uma sequência de pontos consecutivos no tempo cujo comportamento conjunto é incomum, embora cada observação individualmente não seja necessariamente um outlier pontual. Os outliers de subsequência também podem ser **globais** ou **locais** e podem afetar uma única variável dependente do tempo (**outlier de subsequência univariado**) ou múltiplas variáveis dependentes do tempo (**outlier de subsequência multivariado**).



##Algoritmos de Detecção de Anomalias

A **detecção de anomalias** é o processo de identificar pontos de dados que se desviam dos padrões esperados em um conjunto de dados. Diversas aplicações, incluindo **detecção de fraudes**, **detecção de intrusões** e **detecção de falhas**, utilizam técnicas de detecção de anomalias.

O objetivo da detecção de anomalias é encontrar eventos incomuns ou muito raros que possam indicar um **risco potencial**, um **problema** ou até mesmo uma **oportunidade**. Esses eventos se diferenciam significativamente do comportamento normal observado nos dados e, por isso, merecem atenção e investigação adicionais.

### Séries Temporais e Detecção de Anomalias

No contexto de **séries temporais**, os algoritmos de detecção de anomalias são especialmente importantes, pois ajudam a identificar padrões incomuns nos dados que não seriam facilmente percebidos apenas pela observação dos valores brutos.

As anomalias em séries temporais podem se manifestar de diversas formas, tais como:

* Aumentos abruptos nos valores;
* Quedas repentinas;
* Padrões incomuns de comportamento;
* Sazonalidades inesperadas.

Uma série temporal é uma coleção de observações registradas ao longo do tempo.

Os dados de séries temporais podem ser utilizados para treinar algoritmos de detecção de anomalias, como os **Autoencoders**, permitindo que aprendam a representar os padrões típicos do comportamento normal. Posteriormente, essa representação é utilizada para identificar comportamentos anormais.

Essa abordagem consiste em treinar um Autoencoder com dados normais da série temporal para que ele aprenda uma versão compactada dos dados. Em seguida, calcula-se uma **pontuação de anomalia (anomaly score)** por meio do **erro de reconstrução**, que corresponde à diferença entre os dados originais e os dados reconstruídos pelo modelo.

Assim, observações que apresentam erros de reconstrução elevados são consideradas potenciais anomalias.

Os algoritmos de detecção de anomalias aplicados a séries temporais permitem identificar padrões incomuns que podem indicar riscos, problemas ou oportunidades. Por exemplo, no contexto de **manutenção preditiva**, uma anomalia em uma série temporal pode sinalizar uma possível falha de equipamento antes que ela provoque longos períodos de indisponibilidade ou gere riscos à segurança.

Da mesma forma, em aplicações financeiras, anomalias em séries temporais podem revelar movimentos de mercado ou padrões inesperados em previsões financeiras que podem ser explorados para apoiar a tomada de decisões.


###Exemplos de algoritmos para detecção de anomalias
Existem diversos algoritmos para **detecção de anomalias**, cada um mais adequado para determinados tipos de dados e problemas. Alguns dos mais utilizados são:

### 1. Isolation Forest

* Baseado em árvores de decisão.
* Identifica anomalias isolando observações que são mais fáceis de separar dos demais dados.
* Eficiente para grandes volumes de dados.
* Muito utilizado em detecção de fraudes e monitoramento de sistemas.

### 2. Local Outlier Factor (LOF)

* Compara a densidade local de um ponto com a densidade de seus vizinhos.
* Detecta anomalias locais que podem não ser percebidas por métodos globais.
* Adequado para conjuntos de dados com distribuições complexas.

### 3. One-Class SVM

* Aprende o padrão dos dados normais e identifica observações que se afastam desse padrão.
* Funciona bem quando há muitos exemplos normais e poucos ou nenhum exemplo de anomalias.
* Bastante utilizado em segurança cibernética.

### 4. K-Means para Detecção de Anomalias

* Agrupa dados em clusters.
* Pontos muito distantes dos centroides dos grupos podem ser considerados anomalias.
* Simples e fácil de implementar.

### 5. DBSCAN

* Algoritmo de agrupamento baseado em densidade.
* Pontos que não pertencem a nenhuma região densa são classificados como ruído (anomalias).
* Eficiente para detectar outliers em dados espaciais.

### 6. Autoencoders

* Redes neurais treinadas para reconstruir os dados de entrada.
* Dados normais são reconstruídos com baixo erro.
* Anomalias apresentam erro de reconstrução elevado.
* Muito utilizado em imagens, sensores IoT e séries temporais.

### 7. Redes LSTM para Séries Temporais

* Capturam dependências temporais de longo prazo.
* Detectam anomalias comparando valores observados com valores previstos.
* Aplicadas em monitoramento industrial, energia e finanças.

### 8. Prophet + Análise de Resíduos

* Utiliza previsões de séries temporais para identificar desvios significativos.
* Muito empregado em monitoramento de métricas de negócio.

### 9. Métodos Estatísticos (Z-Score e IQR)

#### Z-Score

* Mede quantos desvios-padrão um valor está distante da média.
* Regra comum: |Z| > 3 indica possível anomalia.

#### IQR (Intervalo Interquartil)

* Define limites:

  * Limite inferior = Q1 − 1,5 × IQR
  * Limite superior = Q3 + 1,5 × IQR
* Valores fora desses limites são considerados outliers.

### 10. ARIMA para Séries Temporais

* Modela o comportamento esperado da série.
* Anomalias são identificadas pelos resíduos (diferença entre valor observado e previsto).
* Muito utilizado em previsão e monitoramento temporal.

### Comparação rápida

| Algoritmo        | Tipo de Dados                           | Vantagem Principal                 |
| ---------------- | --------------------------------------- | ---------------------------------- |
| Isolation Forest | Tabulares e temporais                   | Rápido e escalável                 |
| LOF              | Tabulares                               | Detecta anomalias locais           |
| One-Class SVM    | Tabulares                               | Bom para dados normais sem rótulos |
| DBSCAN           | Espaciais e clusters                    | Detecta ruído naturalmente         |
| Autoencoder      | Grandes volumes e alta dimensionalidade | Captura padrões complexos          |
| LSTM             | Séries temporais                        | Considera dependências temporais   |
| ARIMA            | Séries temporais                        | Simples e interpretável            |
| Z-Score / IQR    | Dados estatísticos                      | Fácil implementação                |

Para **séries temporais**, os algoritmos mais utilizados são **Isolation Forest**, **ARIMA**, **LSTM**, **Autoencoders** e métodos estatísticos como **Z-Score** e **IQR**, dependendo da complexidade dos dados e da necessidade de interpretação dos resultados.


##Autoencoders para Detecção de Anomalias em Séries Temporais

Os **Autoencoders** são um tipo de rede neural amplamente utilizado para **detecção de anomalias**, especialmente quando os dados apresentam padrões complexos que métodos estatísticos tradicionais têm dificuldade em capturar.

A detecção de anomalias com Autoencoders baseia-se na ideia de que uma rede neural treinada apenas com comportamentos normais será capaz de reconstruí-los com baixo erro. Quando surge um padrão incomum na série temporal, o erro de reconstrução aumenta significativamente, sinalizando uma possível anomalia. Por essa razão, Autoencoders são uma das técnicas mais populares para monitoramento de equipamentos, sistemas industriais, redes de computadores, finanças e aplicações de Internet das Coisas (IoT).

#### Conceito Básico

Um Autoencoder é treinado para **aprender a reconstruir seus próprios dados de entrada**. Em outras palavras, ele recebe uma informação como entrada, comprime essa informação em uma representação mais compacta e depois tenta reconstruí-la da forma mais fiel possível.

A arquitetura é composta por duas partes principais:

* **Encoder (Codificador):** reduz a dimensionalidade dos dados, extraindo as características mais importantes.
* **Decoder (Decodificador):** reconstrói os dados originais a partir da representação compactada.

Fluxo simplificado:

```
Entrada → Encoder → Representação Compacta → Decoder → Reconstrução
```

#### Como funciona na detecção de anomalias?

A ideia central é treinar o Autoencoder utilizando apenas dados considerados normais.

Durante o treinamento, a rede aprende os padrões, tendências e comportamentos típicos da série temporal. Como consequência, ela consegue reconstruir muito bem exemplos semelhantes aos que viu durante o treinamento.

Quando um dado anômalo é apresentado:

* O Autoencoder não consegue reconstruí-lo adequadamente;
* O erro entre o valor original e o valor reconstruído torna-se elevado;
* Esse erro é utilizado como indicador de anomalia.

#### Exemplo intuitivo

Imagine uma série temporal que registra a temperatura de um equipamento industrial:

| Tempo | Temperatura (°C) |
| ----- | ---------------- |
| 1     | 70               |
| 2     | 72               |
| 3     | 71               |
| 4     | 73               |
| 5     | 120              |

Após ser treinado com temperaturas normais entre 65°C e 75°C, o Autoencoder consegue reconstruir valores próximos de:

```
Entrada:      70 72 71 73
Reconstrução: 69 73 71 72
```

O erro é pequeno.

Entretanto, para uma observação anômala:

```
Entrada:      120
Reconstrução: 74
```

O erro é muito grande, indicando um comportamento incomum.

#### Erro de Reconstrução

A métrica mais utilizada é o **Erro Quadrático Médio (MSE – Mean Squared Error)**:

$$
MSE = \frac{1}{n}\sum_{i=1}^{n}(x_i-\hat{x}_i)^2
$$

onde:

* $(x_i)$ = valor real;
* $(\hat{x}_i)$ = valor reconstruído pelo Autoencoder.

Quanto maior o MSE, maior a probabilidade de a observação ser uma anomalia.

#### Aplicação em Séries Temporais

Nas séries temporais, normalmente não se utiliza apenas um ponto isolado, mas sim uma **janela temporal**.

Exemplo:

```
Janela 1:
[70, 71, 72, 71, 73]

Janela 2:
[71, 72, 71, 73, 120]
```

O Autoencoder aprende o padrão normal das janelas e identifica aquelas cuja reconstrução apresenta erro elevado.

Isso permite detectar:

* Picos inesperados;
* Quedas abruptas;
* Mudanças de regime;
* Falhas em sensores;
* Comportamentos operacionais anormais.

#### Vantagens

✅ Não exige dados rotulados (aprendizado não supervisionado).

✅ Capaz de modelar relações não lineares complexas.

✅ Funciona bem em séries temporais de alta dimensionalidade.

✅ Pode detectar anomalias que não seriam encontradas por métodos estatísticos simples.

#### Limitações

❌ Requer quantidade razoável de dados normais para treinamento.

❌ Treinamento mais custoso computacionalmente.

❌ A escolha do limiar de erro para definir uma anomalia pode ser desafiadora.

❌ Pode aprender a reconstruir algumas anomalias se elas estiverem presentes em excesso no conjunto de treinamento.

## Detecção de Anomalias (Outliers) em Séries Temporais com LSTM

Os **Autoencoders** são uma técnica de aprendizado não supervisionado (*unsupervised learning*), embora sejam treinados utilizando métodos de aprendizado supervisionado. Seu objetivo é minimizar o **erro de reconstrução** com base em uma função de perda, como o **Erro Quadrático Médio (Mean Squared Error – MSE)**.

Neste exemplo, buscaremos detectar anomalias na série temporal histórica dos preços das ações da **Johnson & Johnson** utilizando um **Autoencoder baseado em LSTM (Long Short-Term Memory)**. Os dados podem ser obtidos no Yahoo Finance.

O período selecionado para análise compreende **02/01/1985 a 25/06/2024**.

As etapas para detectar anomalias nos preços das ações da Johnson & Johnson utilizando um Autoencoder LSTM são as seguintes:

1. **Treinar um Autoencoder LSTM** utilizando os dados históricos das ações da Johnson & Johnson no período de **02/01/1985 a 01/03/2017**. Assume-se que os dados desse intervalo não contêm anomalias e representam o comportamento normal da série temporal.

2. **Utilizar o Autoencoder LSTM treinado** para reconstruir os dados do período de teste, compreendido entre **01/03/2017 e 25/06/2024**, calculando o erro de reconstrução para cada observação.

3. **Identificar anomalias** comparando o erro de reconstrução com um valor limite (*threshold*). Caso o erro de reconstrução de uma observação seja superior ao limite estabelecido, essa observação será classificada como uma anomalia.

Para compreender melhor o funcionamento dessa abordagem, iremos analisar a arquitetura de um **Autoencoder LSTM**, examinando cada camada da rede neural e sua função no processo de aprendizado e reconstrução dos dados.

###Create sequences
Convert input data into 3-D array combining TIME_STEPS. The shape of the array should be [samples, TIME_STEPS, features], as required for LSTM network.

We want our network to have memory of 30 days, so we set TIME_STEPS=30.

````python
TIME_STEPS=30

def create_sequences(X, y, time_steps=TIME_STEPS):
    Xs, ys = [], []
    for i in range(len(X)-time_steps):
        Xs.append(X[i:(i+time_steps)])
        ys.append(y[i+time_steps])

    return np.array(Xs), np.array(ys)

X_train, y_train = create_sequences(train_sc, train_sc.reshape(-1))
X_test, y_test = create_sequences(test_sc, test_sc.reshape(-1))

print(f'Training shape: {X_train.shape}')
print(f'Testing shape: {X_test.shape}')
````

### Construção do Modelo

Definimos uma arquitetura de **Autoencoder LSTM de reconstrução** que recebe como entrada sequências contendo **30 passos de tempo (time steps)** e **uma única característica (feature)**, produzindo como saída uma sequência com a mesma estrutura: **30 passos de tempo e uma característica**.

A camada **RepeatVector(30)** é utilizada para repetir a representação codificada (vetor latente) **30 vezes**, permitindo que o decodificador gere novamente toda a sequência temporal.

O parâmetro **return_sequences=True** é configurado nas camadas LSTM do decodificador para que a saída continue sendo uma sequência completa, em vez de retornar apenas o último estado temporal. Dessa forma, o modelo consegue reconstruir todos os 30 passos da sequência de entrada.

Por fim, adiciona-se a camada:

```python
TimeDistributed(Dense(X_train.shape[2]))
```

Essa camada é responsável por gerar a saída final do modelo. A função **TimeDistributed** aplica a mesma camada **Dense** a cada passo temporal da sequência. O parâmetro **X_train.shape[2]** corresponde ao número de características (*features*) presentes nos dados de entrada. Como neste exemplo existe apenas uma variável (o preço da ação), a saída terá uma característica para cada um dos 30 passos temporais reconstruídos.

Em resumo, a arquitetura funciona da seguinte forma:

```text
Sequência de Entrada (30 x 1)
            ↓
        Encoder LSTM
            ↓
      Espaço Latente
            ↓
      RepeatVector(30)
            ↓
       Decoder LSTM
            ↓
TimeDistributed(Dense)
            ↓
Sequência Reconstruída (30 x 1)
```

O objetivo do treinamento é fazer com que a sequência reconstruída seja o mais próxima possível da sequência original. Posteriormente, o **erro de reconstrução** será utilizado para identificar anomalias: sequências com erro elevado serão consideradas comportamentos anormais.


````python
# Limpa a sessão atual do Keras/TensorFlow para evitar conflitos
# com modelos criados anteriormente e liberar memória.
K.clear_session()

# Cria um modelo sequencial, onde as camadas são empilhadas
# uma após a outra.
model = Sequential()

# Define a camada de entrada.
# X_train.shape[1] = número de passos temporais (time steps)
# X_train.shape[2] = número de variáveis (features)
model.add(Input((X_train.shape[1], X_train.shape[2])))

# Camada Encoder LSTM
# Comprime a sequência de entrada em uma representação latente
# contendo 128 unidades (neurônios).
model.add(LSTM(128))

# Dropout de 20%
# Durante o treinamento, desativa aleatoriamente 20% dos neurônios
# para reduzir overfitting e melhorar a generalização do modelo.
model.add(Dropout(rate=0.2))

# Repete o vetor latente gerado pelo Encoder
# pelo mesmo número de passos temporais da sequência original.
#
# Essa etapa prepara os dados para o Decoder reconstruir
# toda a sequência temporal.
model.add(RepeatVector(X_train.shape[1]))

# Camada Decoder LSTM
# Reconstrói a sequência temporal a partir do vetor latente.
#
# return_sequences=True faz com que a LSTM retorne uma saída
# para cada instante da sequência, e não apenas o último estado.
model.add(LSTM(128, return_sequences=True))

# Novo Dropout para reduzir overfitting durante a reconstrução.
model.add(Dropout(rate=0.2))

# Aplica uma camada Dense a cada passo temporal da sequência.
#
# X_train.shape[2] corresponde ao número de variáveis
# da série temporal.
#
# Se a série possui apenas uma variável (ex.: preço da ação),
# a saída terá uma dimensão por instante de tempo.
model.add(TimeDistributed(Dense(X_train.shape[2])))

# Compila o modelo.
#
# optimizer='adam'
# Utiliza o algoritmo Adam para ajustar os pesos da rede.
#
# loss='mae'
# Utiliza o Erro Absoluto Médio (Mean Absolute Error)
# como função de perda para medir a diferença entre
# a sequência original e a sequência reconstruída.
model.compile(
    optimizer='adam',
    loss='mae'
)

# Exibe um resumo da arquitetura do modelo,
# mostrando camadas, dimensões de saída e número de parâmetros.
model.summary()
````

````python
history = model.fit(X_train, y_train, epochs=100, batch_size=32, validation_split=0.1,
                    callbacks=[keras.callbacks.EarlyStopping(monitor='val_loss', patience=3, mode='min')], shuffle=False)
````

````python
# Returns the loss value & metrics values for the model in test mode
model.evaluate(X_test, y_test)
````

###Determine Anomalies
Find MAE loss on the training data.

Make the max MAE loss value in the training data as the reconstruction error threshold.

If the reconstruction loss for a data point in the test set is greater than this reconstruction error threshold value then we will label this data point as an anomaly.

````python
X_train_pred = model.predict(X_train, verbose=0)
train_mae_loss = np.mean(np.abs(X_train_pred - X_train), axis=1)

plt.hist(train_mae_loss, bins=50)
plt.xlabel('Train MAE loss')
plt.ylabel('Number of Samples');

threshold = np.max(train_mae_loss)
print(f'Reconstruction error threshold: {threshold}')
````

````python
X_test_pred = model.predict(X_test, verbose=0)
test_mae_loss = np.mean(np.abs(X_test_pred-X_test), axis=1)

plt.hist(test_mae_loss, bins=50)
plt.xlabel('Test MAE loss')
plt.ylabel('Number of samples');
````

````python
test_score_df = pd.DataFrame(test[TIME_STEPS:])
test_score_df['loss'] = test_mae_loss
test_score_df['threshold'] = threshold
test_score_df['anomaly'] = test_score_df['loss'] > test_score_df['threshold']
test_score_df['Close'] = test[TIME_STEPS:]['Close']

fig = go.Figure()
fig.add_trace(go.Scatter(x=test_score_df['Date'], y=test_score_df['loss'], name='Test loss'))
fig.add_trace(go.Scatter(x=test_score_df['Date'], y=test_score_df['threshold'], name='Threshold'))
fig.update_layout(showlegend=True, title='Test loss vs. Threshold')
fig.show()
````

````python
anomalies = test_score_df.loc[test_score_df['anomaly'] == True]
anomalies.shape
````

As you can see, there are 2 data points in the test set that exceeded the reconstruction error threshold.

````python
plt.plot(test_score_df['Date'], test_score_df['Close'], '.')
plt.plot(anomalies['Date'], anomalies['Close'], 'o')
plt.title('Anomalies')
plt.show()
````
